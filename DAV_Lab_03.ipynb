```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Movie": [
        "Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
        "Movie F", "Movie G", "Movie H", "Movie I", "Movie J",
        "Movie K", "Movie L", "Movie M", "Movie N", "Movie O"
    ],
    "Rating": [
        7.8, 8.2, 6.9, 7.5, 8.7,
        6.5, 7.1, 8.9, 7.6, 6.8,
        8.4, 7.3, 9.0, 6.7, 8.1
    ],
    "Budget_Crore": [
        50, 80, 30, 60, 120,
        25, 45, 150, 70, 35,
        100, 55, 180, 40, 90
    ],
    "Revenue_Crore": [
        120, 200, 70, 140, 350,
        55, 100, 450, 180, 75,
        280, 130, 500, 90, 230
    ],
    "Runtime_Minutes": [
        125, 140, 110, 130, 150,
        105, 120, 155, 135, 115,
        145, 128, 160, 108, 138
    ]
}

df = pd.DataFrame(data)

df
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Rating"], bins=6, edgecolor="black")

plt.title("Histogram of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Budget_Crore"])

plt.title("Box Plot of Movie Budget")
plt.ylabel("Budget (Crore)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Budget_Crore"], df["Revenue_Crore"])

plt.title("Budget vs Revenue")
plt.xlabel("Budget (Crore)")
plt.ylabel("Revenue (Crore)")

plt.show()
```

```python
numerical_columns = [
    "Rating",
    "Budget_Crore",
    "Revenue_Crore",
    "Runtime_Minutes"
]

sns.pairplot(df[numerical_columns])

plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = df[numerical_columns].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
print("1. Most movie ratings are concentrated between 7 and 8.5, showing that the majority of movies have moderate to high ratings.")
print("2. Movies with higher budgets generally have higher revenues, showing a positive relationship between Budget_Crore and Revenue_Crore.")
print("3. Revenue_Crore has a stronger relationship with Budget_Crore than with Rating, indicating that movie budget is more closely associated with revenue in this dataset.")
```
